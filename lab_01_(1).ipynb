##5: Text Analysis Using NumPy (NLP Track)

**Demonstration** :


```python
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"
tokens2 = sentence2.lower().split()
print(tokens2)
```

**Explanation** : We convert first the sentence to lowercase and then split it into individual words, then store those words in tokens2

**Building a Vocabulary** :

```python
tokens_array = np.array(tokens2)
vocabulary = np.unique(tokens_array)

print("Vocabulary:", vocabulary)
print("Vocabulary size:", len(vocabulary))
```

**Explanation** : I converted the words into a NumPy array. Then, np.unique() finds the words that are not repeated. Finally, I print those unique words and count them. Here, we have 11 unique words.

**Word Frequency Vector (Bag-of-Words)** :


**Predection** : The word that will have a count of 2 is:

numpy

All the other words appear only once, so their count will be 1.

```python
def bag_of_words_vector(tokens, vocabulary):
    vector = np.array([np.sum(np.array(tokens) == word)
                       for word in vocabulary])
    return vector

tokens2_array = np.array(tokens2)
vocabulary2 = np.unique(tokens2_array)
vec2 = bag_of_words_vector(tokens2, vocabulary2)

print("Vocabulary:", vocabulary2)
print("Vector:", vec2)
```

**Explanation** : I am creating a Bag of Words vector. It checks each word in the vocabulary and counts how many times that word appears in the sentence. Since every word in tokens2 appears only once, the vector contains 1 for each word

**Comparing Two Sentences (Cosine Similarity)** :

```python
sentence1 = "I like playing football"
sentence2 = "I enjoy playing football"

sentence3 = "I like playing football"
sentence4 = "The weather is very cold today"

def cosine_similarity(sentence_a, sentence_b):

    tokens_a = sentence_a.lower().split()
    tokens_b = sentence_b.lower().split()

    vocabulary = np.unique(np.array(tokens_a + tokens_b))

    vec_a = bag_of_words_vector(tokens_a, vocabulary)
    vec_b = bag_of_words_vector(tokens_b, vocabulary)

    dot_product = np.dot(vec_a, vec_b)

    norm_a = np.linalg.norm(vec_a)
    norm_b = np.linalg.norm(vec_b)

    similarity = dot_product / (norm_a * norm_b)

    return similarity

similar_score = cosine_similarity(sentence1, sentence2)
different_score = cosine_similarity(sentence3, sentence4)

print("Similarity score:", similar_score)
print("Different score:", different_score)
```

**explanation** :
The first pair talks about playing football, so they should have a higher similarity score.
The second pair talks about football and weather, so they should have a lower similarity score.

**Reflection** : Cosine similarity tells us how similar two sentences are.
If two sentences have similar words, their value is close to 1.
If two sentences are very different, their value is close to 0.
A cosine similarity of exactly 1 means the two vectors are completely similar and point in the same direction.

```python
text = "Python is easy to learn. Python is useful for data science. Data science is interesting. Python is useful and powerful."

tokens = text.lower().split()
tokens_array = np.array(tokens)
vocabulary = np.unique(tokens_array)
vector = bag_of_words_vector(tokens, vocabulary)

sorted_indices = np.argsort(vector)
top_3_indices = sorted_indices[::-1][:3]

for i in top_3_indices:
    print(vocabulary[i], ":", vector[i])
```

### #Mini Project: Simple Text Similarity Checker (NLP Track)

```python
import numpy as np

sentence1 = "I love playing football"
sentence2 = "I enjoy playing football"
sentence3 = "The weather is very cold today"

tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

all_tokens = tokens1 + tokens2 + tokens3
vocabulary = np.unique(np.array(all_tokens))

def bag_of_words_vector(tokens, vocabulary):
    vector = np.array([np.sum(np.array(tokens) == word)
                       for word in vocabulary])
    return vector

vec1 = bag_of_words_vector(tokens1, vocabulary)
vec2 = bag_of_words_vector(tokens2, vocabulary)
vec3 = bag_of_words_vector(tokens3, vocabulary)

def cosine_similarity(vec_a, vec_b):
    dot_product = np.dot(vec_a, vec_b)
    norm_a = np.linalg.norm(vec_a)
    norm_b = np.linalg.norm(vec_b)

    return dot_product / (norm_a * norm_b)

similarity_12 = cosine_similarity(vec1, vec2)
similarity_13 = cosine_similarity(vec1, vec3)
similarity_23 = cosine_similarity(vec2, vec3)

print("Sentence 1 and 2:", similarity_12)
print("Sentence 1 and 3:", similarity_13)
print("Sentence 2 and 3:", similarity_23)

scores = [similarity_12, similarity_13, similarity_23]
highest_score = max(scores)

if highest_score == similarity_12:
    print("Most similar: Sentence 1 and Sentence 2")

elif highest_score == similarity_13:
    print("Most similar: Sentence 1 and Sentence 3")

else:
    print("Most similar: Sentence 2 and Sentence 3")
```

**Explanation** : I take three sentences. Then I split each sentence into individual words. I create one common vocabulary containing all the unique words from the three sentences. After that, I convert each sentence into a Bag-of-Words vector, where the numbers represent how many times each word appears. Then I calculate cosine similarity between every pair of sentences. Finally, I compare the three similarity scores and print the pair with the highest score as the most similar sentences.
